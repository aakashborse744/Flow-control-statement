```python
# The table should show the number given by the user.
num = int(input("Enter a number: "))

count = 1
while count <= 10:       #The condition here is i <= 9, meaning "keep looping as long as i is less than or equal to 9."
    print(num, "*", count, "=", num * count)
    count+= 1
```

```python
# # print full table for numberrs 2 through 9 (Not just one number), each showing one to 9.
num = eval(input("Enter the number:"))
while num <= 10:    # Nested while loop
    i = 1
    while i <= 10:
        print(num, "*", i, "=", num * i)
        i += 1
    print()  # blank line between tables
    num += 1
```

```python
# Diamond pattern: Take an odd number n and print a diamond shape made 
# of * using nested for loops (upper triangle expanding, then lower triangle shrinking).

n = int(input("Enter an odd number: "))

if n % 2 == 0:
    print("n must be odd")
else:
    mid = n // 2

    # Upper half (including middle row)
    for i in range(mid + 1):
        # print spaces
        for j in range(mid - i):
            print(" ", end=" ")
        # print stars
        for k in range(2 * i + 1):
            print("*", end=" ")
        print()

    # Lower half
    for i in range(mid - 1, -1, -1):
        # print spaces
        for j in range(mid - i):
            print(" ", end=" ")
        # print stars
        for k in range(2 * i + 1):
            print("*", end=" ")
        print()
```

```python
#3 . Hollow Square
n =eval(input("Enter the number:"))
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*" , end=" ")
        else:
            print(" " , end=" ")
    print()
```

# 🔴 Hard (Multiple conditions / combined logic)

```python
#4. Butterfly Pattern
n = eval(input("Enter the number : "))

for i in range(0 , n + 1):
    for j in range(i):
        print("*" , end=" ")
    for j in range(2 * ( n - i)):
        print(" " , end=" ")
    for j in range(i):
        print("*" , end=" ")
    print()
```
